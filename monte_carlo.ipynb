```python
CARDS = [2, 3, 4, 5, 6, 7, 8, 9, 10, 10, 10, 10, 11]  # 11 represents an Ace
cards_length = len(CARDS)

import random
def draw_card():
    return random.choice(CARDS)


def dealer_total(dealer_upcard):
    ace = dealer_upcard == 11
    total = dealer_upcard
    while total < 17:
        card = draw_card()
        total += card

        if card == 11:
            if total > 21:
                total -= 10
            else:
                ace = True

        if total > 21 and ace:
            total -= 10
            ace = False
    return total



from collections import defaultdict
table = defaultdict(int)


for i in range(1, 12):
    for _ in range(1000000):
        dealer_upcard = i
        total = dealer_total(dealer_upcard)
        if total > 21:
            total = 22
        table[(dealer_upcard, total)] += 1


random.seed(42)
print("Dealer Upcard | Dealer Total | Count")
for (dealer_upcard, total), count in sorted(table.items()):
    table[(dealer_upcard, total)] = count / 1000000  # Convert to probability
    probability = table[(dealer_upcard, total)]
    print(f"{dealer_upcard:14} | {total:12} | {100*probability:.4f}%")
```

```python
table = defaultdict(lambda: (0, 0, 0, 0, 0, 0))
table[22] = (0, 0, 0, 0, 0, 1)
table[21] = (0, 0, 0, 0, 1, 0)
table[20] = (0, 0, 0, 1, 0, 0)
table[19] = (0, 0, 1, 0, 0, 0)
table[18] = (0, 1, 0, 0, 0, 0)
table[17] = (1, 0, 0, 0, 0, 0)



for i in range(16, 2, -1):
    odds = [0, 0, 0, 0, 0, 0]
    for c in CARDS:
        total = i + c
        if total > 21:
            total = 22
        for j in range(6):
            odds[j] += table[total][j]/cards_length

    table[i] = tuple(odds)
    print(f"Total: {i}, Odds: {table[i]}")

```
